```python
import os
import matplotlib.pyplot as plt 
```

```python
### function for creating class count in dataset partitions
def post_dataset_count(input_path):
    classCount = {}
    ### reading all files inn the input folder 
    for file in os.listdir(input_path):
        ### checking that the files is a .txt file
        if file.endswith(('.txt')):
            ### removing file extention
            base, ext = os.path.splitext(file)
            ### striping numbers from file name
            str = base.strip("0123456789")
            ### adding count to class count dict
            if(classCount.get(str)):
                classCount[str] += 1
            else:
                classCount[str] = 1
    return classCount

### function for creating class count for full dataset            
def pre_dataset_count(input_path):
    classCount = {}
    ### finding all dirs in the input path
    for root, dirs, file in os.walk(input_path):
        ### looping all dirs
        for d in dirs:
            ### checking that d is not the label dir
            if d != "Label":
                ### creatring the class in the output dict
                classCount[d] = 0
                d_path = os.path.join(root, d)
                ### looping all images files in d to count the images for the class.
                for file in os.listdir(d_path):
                    if file.endswith(('.jpg', '.jpeg', '.png', '.bmp')):
                        classCount[d] += 1
    return classCount
```

```python
### runing functions to create class counts for the dataset and partitions
pre_count = pre_dataset_count("./data/pre")
post_train_count = post_dataset_count("./data/post/1/labels/train")
post_val_count = post_dataset_count("./data/post/1/labels/val")
post_test_count = post_dataset_count("./data/post/1/labels/test")
```

```python
### creating output images for class count for full dataset.
plt.figure(figsize=(20,10))
plt.bar(pre_count.keys(), pre_count.values())
plt.title("Full dataset class count")
plt.xlabel("Class label")
plt.ylabel("Count")
plt.show()
```

```python
### creating output images for class count for train partitions in subsets
plt.figure(figsize=(20,10))
plt.bar(post_train_count.keys(), post_train_count.values())
plt.title("Subset train partition class count")
plt.xlabel("Class label")
plt.ylabel("Count")
plt.show()
```

```python
### creating output images for class count for val partitions in subsets
plt.figure(figsize=(20,10))
plt.bar(post_val_count.keys(), post_val_count.values())
plt.title("Subset val partition class count")
plt.xlabel("Class label")
plt.ylabel("Count")
plt.show()
```

```python
### creating output images for class count for test partitions in subsets
plt.figure(figsize=(20,10))
plt.bar(post_test_count.keys(), post_test_count.values())
plt.title("Subset test partition class count")
plt.xlabel("Class label")
plt.ylabel("Count")
plt.show()
```
